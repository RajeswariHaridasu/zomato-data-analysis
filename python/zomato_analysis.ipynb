# Zomato Bengaluru Restaurant Analysis

End-to-end exploratory data analysis using **Python and Pandas**. The project examines restaurant ratings, votes, online ordering, table booking, restaurant types, cuisines, locations, and approximate cost for two people.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('../data/zomato_sample.csv')
df.head()
```

## Data Quality Checks

```python
df.info()
df.isna().sum().sort_values(ascending=False).head(10)
```

## Cleaning

```python
# Clean rating values such as '4.1/5' and 'NEW'
df['rating'] = pd.to_numeric(df['rate'].astype(str).str.extract(r'(\d+(?:\.\d+)?)')[0], errors='coerce')

# Convert votes and cost to numeric
df['votes'] = pd.to_numeric(df['votes'], errors='coerce')
df['cost_for_two'] = pd.to_numeric(
    df['approx_cost(for two people)'].astype(str).str.replace(',', '', regex=False),
    errors='coerce'
)

df = df.drop_duplicates(subset=['name', 'location', 'cuisines'])
df[['name','rating','votes','cost_for_two']].head()
```

## Key Business Questions

```python
# Top locations by restaurant count
location_summary = df.groupby('location').size().sort_values(ascending=False).head(10)
location_summary
```

```python
# Restaurant types with the highest average rating
type_summary = (df.dropna(subset=['rating'])
                .groupby('rest_type')
                .agg(restaurant_count=('name','count'), avg_rating=('rating','mean'))
                .sort_values('avg_rating', ascending=False)
                .head(10))
type_summary
```

```python
# Online-order adoption
online_order_summary = df['online_order'].value_counts(dropna=False)
online_order_summary
```

## Visualization

```python
location_summary.sort_values().plot(kind='barh', figsize=(8,5), title='Top Locations by Restaurant Count')
plt.xlabel('Restaurant Count')
plt.tight_layout()
plt.show()
```

```python
df.dropna(subset=['rating']).plot.scatter(x='rating', y='votes', alpha=0.35, figsize=(8,5), title='Rating vs Votes')
plt.tight_layout()
plt.show()
```

## Key Takeaways
- Identify locations with the highest restaurant concentration.
- Compare restaurant types by average rating and review volume.
- Measure online-order and table-booking adoption.
- Explore the relationship between ratings, votes, and approximate customer spend.
